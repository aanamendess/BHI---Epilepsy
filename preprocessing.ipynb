## Dataset creation
Loop through the paths for the signals and create a dataset with the raw segment files and the description containing subject, path, target and labels for each segment.

```python
# Creation of the Dataset

descriptions = []
raws = [] # list containing all the raw signals

for i, path in enumerate(paths):
    raw = read_raw_edf(path, preload=False) # preload = false because we just want to rename the channels, no need to load the data (faster)

    raw.set_meas_date(None)

    path_csv_bi = path.replace("edf", "csv_bi")
                        
    
    desc = {
        "subject": raw.info['subject_info'],
        "path": path,
        "target": 1 if i < len(signals_epi) else 0,  # epilepsy vs non
        "labels": pd.read_csv(path_csv_bi, comment='#')

    }

    descriptions.append(desc)
    raws.append(raw)

base_datasets = [
    BaseDataset(raw, description=desc, target_name='target')
    for raw, desc in zip(raws, descriptions)
]

tuh = BaseConcatDataset(base_datasets)

```

```python
# Standardize all the channels names and channel types

def rename_channels(ds):

    filtered_datasets = []
    #montage = make_standard_montage("standard_1020")
    
    for d in ds.datasets:
        signal = d.raw
        names = signal.ch_names[:]
        signal.rename_channels(mapping=lambda x: x.replace('EEG ', '').replace('-LE', '').replace('-REF', '').replace('EKG1', 'EKG').replace('PHOTIC PH', 'PHOTIC'))
        
        #signal.set_montage(montage, on_missing="ignore")

        if 'EKG' in signal.ch_names[:] and 'PZ' in signal.ch_names[:]:
            if 'PHOTIC' not in signal.ch_names[:]:
                signal.set_channel_types(mapping={'EKG': 'ecg'})
                signal.pick(channel_names)
            else:
                signal.set_channel_types(mapping={'EKG': 'ecg', 'PHOTIC':'stim'})
                signal.pick(channel_names_ph)
                

            filtered_datasets.append(d)
    
    return BaseConcatDataset(filtered_datasets)

tuh = rename_channels(tuh)
```

## Functions to select relevant segments (segments with at least 5 minutes).

```python
def crop_first(raw, tmin, raw_path):
    """ Crops first minute of each session. """
    
    tmax = raw.times[-1]
    if 't000' in raw_path:
        return raw.copy().crop(tmin=tmin, tmax=tmax)
    return raw
```

```python

def duration_seizure(dataset):
    """ Checks start of seizure and crops the segment to end 5 minutes before start. """
    
    raw_cropped_seiz = dataset.raw
    annotations = dataset.description['labels']

    for i, l in enumerate(annotations['label']):
        if l == 'seiz':
            start_time = annotations['start_time'].iloc[i]
            tmax = (start_time - 5 * 60)
            if tmax < 0:
                tmax = 0
            raw_cropped_seiz = raw_cropped_seiz.copy().crop(tmin=0, tmax=tmax) 
            break
        
    return raw_cropped_seiz
```

```python
def duration_photic(raw):
    """ Checks start of first photic stimulation and crops segment to end before the first photic starts. """
 
    if 'PHOTIC' in raw.ch_names:
        sfreq = raw.info['sfreq']
        ch_idx = raw.ch_names.index('PHOTIC')
        photic_signal = raw.get_data(picks=[ch_idx])[0]
        if np.all(photic_signal == 0):
            duration = raw.times[-1]
        else:
            first_photic = np.nonzero(photic_signal)[0][0]
            duration = (first_photic / sfreq)
        raw = raw.pick(channel_names)
    else:
        duration = raw.times[-1]

    

    raw_cropped = raw.copy().crop(tmin=0, tmax=duration)

    return raw_cropped, duration
```

```python
# First, we will do some selection of the recordings based on the duration. We will select those recordings, that have at least five minutes duration.

def select_by_duration(ds, tmin):
    """ Loops through all segments and keeps only the ones with at least 5 minutes. """

    # determine length of the recordings and select based on tmin and tmax
    split_ids = []
    for d_i, d in enumerate(ds.datasets):
        raw = d.raw
        if raw.times[-1] >= (tmin + 60):
            raw_cropped_seiz = duration_seizure(d)
            if raw_cropped_seiz.times[-1] > 60:
                # Remove first minute if is first segment
                raw_first = crop_first(raw_cropped_seiz, 60, raw_path = d.description['path'])
                raw_cropped, duration = duration_photic(raw_first)
                
                
                if tmin <= duration <= np.inf:
                    ds.datasets[d_i].raw = raw_cropped
                    split_ids.append(d_i)
    splits = ds.split(split_ids)
    split = splits['0']
    return split


tmin = 5 * 60

tuh = select_by_duration(tuh, tmin)
```

## Preprocessing Functions

```python

def apply_ica(raw):
    """ Apply ICA. """

    ica = mne.preprocessing.ICA(n_components = None, method="infomax", fit_params=dict(extended=True), max_iter=800, random_state=97)
    ica.fit(raw)
    eog_indices, _ = ica.find_bads_eog(raw, ch_name=['FP1', 'FP2'], threshold=3.0)
    ica.exclude = eog_indices
    ica.apply(raw)
    print(type(raw))

```

```python
def clip_fn(raw):
    """ Remove amplitudes outside -800 to 800 microvolts. """

    data = raw.get_data()
    data = np.clip(data, -800, 800)
    raw._data = data
```

```python
def detect_bad_ecg(raw, ecg_ch_name):
    """ Detect bad ECG channels by checking if BPMs are in normal range (40 - 180). """

    ecg = raw.get_data(picks=[ecg_ch_name]).squeeze()
    fs = raw.info['sfreq']
   
    try:
        signals, info = nk.ecg_process(ecg, sampling_rate=fs)
        peaks = info["ECG_R_Peaks"]
       
        if len(peaks) < 5:
            return True
       
        hr = nk.ecg_rate(peaks, sampling_rate=fs)
       
        if np.isnan(hr).all():
            return True
       
        if hr.mean() < 40 or hr.mean() > 180:
            return True
       
        return False
   
    except:
        return True
```

```python
def bad_channel_rejection(raw):
    """ Reject bad segments """

     # EEG
    raw_eeg = raw.copy().pick_types(eeg=True)
    nc = NoisyChannels(
        raw_eeg,
        ransac=False,
        correlation=False,
        do_detrend=True
    )
    nc.find_all_bads()
    bad_eeg = nc.get_bads()
    raw_eeg.info["bads"].extend(bad_eeg)
   
    eeg_ratio = len(bad_eeg) / len(raw_eeg.ch_names)
   
    # ECG
    ecg_chs = mne.pick_types(raw.info, ecg=True)
   
    bad_ecg = False
    if len(ecg_chs) > 0:
        ecg_name = raw.ch_names[ecg_chs[0]]
        bad_ecg = detect_bad_ecg(raw, ecg_name)
   
    # Criterion
    if bad_ecg:
        return True   # reject full recording
   
    if eeg_ratio > 0.3:
        return True
   
    return False
```

```python
def set_avg_ref(raw):
    """ Sets average reference for all segments by creating a new virtual reference electrode, correspondent to the average of the current EEG signal. """

    print(type(raw))
    mne.set_eeg_reference(
        raw,
        ref_channels='average',
        ch_type='eeg',
        projection=False,
        copy=False
    )
```

# Apply some pre-processing steps


```python
filters = [
    Preprocessor('notch_filter', freqs=[60], picks='eeg'),
    Preprocessor('filter', l_freq=0.5, h_freq=75., picks='eeg'),
    Preprocessor('filter', l_freq=2., h_freq=40.,picks='ecg')
]
```

```python
preprocessors = [
    Preprocessor(apply_ica, apply_on_array=False),
    Preprocessor(set_avg_ref, apply_on_array=False),
    Preprocessor(lambda data, factor: np.multiply(data, factor), factor=1e6),
    Preprocessor(clip_fn, apply_on_array=False),
    Preprocessor('resample', sfreq=sfreq),
]
```
